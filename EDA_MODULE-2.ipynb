```python
import numpy as np 
import pandas as pd

population= pd.Series({'California': 38332521,'Texas': 25415625,'New work': 24212154,'Florida': 78542315})
print(population)

df1= pd.DataFrame(population,columns=['population'])
print(df1)

df2= pd.DataFrame(np.random.rand(3,2),columns=['foo','bar'],index=['a','b','c'])
print(df2)

df3= pd.DataFrame([{'a':1,'b': 2},{'b':3,'c':4}])
print(df3)

d=[
    {'a':1,'b': 2*i}
    for i in range(10)
]
df4= pd.DataFrame(d)
print(df4)

ind = pd.Index([2,3,5,7,11])
print(ind)
print(ind.size,ind.shape,ind.ndim,ind.dtype)


print(ind[1])
print(ind[::2])
print(ind[1:4:2])

indA = pd.Index([1,3,5,7,9])
indB = pd.Index([2,3,5,7,11])

print(indA & indB)
print(indA | indB)
print(indA ^ indB)
print(indA.intersection(indB))
```
