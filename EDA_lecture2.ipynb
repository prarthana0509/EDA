```python
arr = np.random.randint(0, 20, size=(2, 3))
x= np.random.randint(0, 20, size=(2, 3))
x1=np.vstack([arr,x])
x2=np.hstack([arr,x])
x3=np.concatenate([arr,x],axis=1)
print(x1)
print(x2)
print(x3)
```
