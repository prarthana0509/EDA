```python
import numpy as np
np.random.seed(1)
x=np.random.randint(0,10,size=5)
x.sort()
print(x)
```

```python
i = np.argsort(x)
print(i)
print(x[i])
```

```python
np.random.seed(1)
x=np.random.randint(0,30,size=(4,6))
print(x)
a=np.sort(x,axis=1)
print(a)
b=np.sort(x,axis=0)
print(b)
```

```python
np.random.seed(1)
x=np.random.randint(0,30,size=15)
print(x)
q= np.partition(x,4)
print(q)
x.sort()
print(x)
```

```python
a=np.array(['ram','sam','ray','snow'])
b=np.array([51,67,89,43])
y= np.argsort(b)
c=a[y]
print(c)
```

```python
np.random.seed(1)
x=np.random.randint(0,3,size=(10,2))
dist_sq = np.sum((x[:, np.newaxis, :] - x[np.newaxis, :, :]) ** 2, axis=-1)
t= np.argsort(dist_sq, axis=1)
print(t)
```
